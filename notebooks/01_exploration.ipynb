The following checks whether the link file contains each interaction once or twice:

```python
a, b = df_links.iloc[0].protein1, df_links.iloc[0].protein2
print(a, b)
fwd = df_links[(df_links.protein1 == a) & (df_links.protein2 == b)]
rev = df_links[(df_links.protein1 == b) & (df_links.protein2 == a)]
print("forward:", len(fwd))
print("reverse:", len(rev))
print(fwd)
print(rev)
```

```python
(df_links.protein1 < df_links.protein2).mean()
```

```python
import numpy as np
a = df_links.protein1.to_numpy()
b = df_links.protein2.to_numpy()
swap = a > b
lo = np.where(swap, b, a)
hi = np.where(swap, a, b)
n_rows = len(df_links)
n_pairs = len(pd.DataFrame({"lo": lo, "hi": hi}).drop_duplicates())
print(n_rows, n_pairs, n_rows / n_pairs)
```
